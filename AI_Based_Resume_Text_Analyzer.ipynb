```python
import nltk
import re

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
resume = """
Muththamizh Selvi K S

Computer Science Engineering student with an interest in
Python, Artificial Intelligence, Machine Learning and Cloud Computing.

Education:
Bachelor of Engineering in Computer Science Engineering.

Skills:
Python, Java, HTML, CSS, JavaScript, SQL, Machine Learning,
Artificial Intelligence, Cloud Computing, Git and GitHub.

Experience:
Completed internship in Web Development and UI/UX Designing.

Certifications:
NPTEL Database Management Systems Certificate.
Completed Tata Young Professional Course.

Projects:
Smart Queue Management System, Complaint Management System,
Cloud-Based Student Task Manager and AI-based applications.
"""
```

```python
# Convert resume to lowercase
text = resume.lower()

# -----------------------------
# Skill Detection
# -----------------------------

skills = [
    "python",
    "java",
    "html",
    "css",
    "javascript",
    "sql",
    "machine learning",
    "artificial intelligence",
    "cloud computing",
    "git",
    "github",
    "ui/ux",
    "flask",
    "database"
]

found_skills = []

for skill in skills:
    if skill in text:
        found_skills.append(skill)

# -----------------------------
# Education Detection
# -----------------------------

education_keywords = [
    "bachelor",
    "engineering",
    "computer science",
    "degree",
    "college",
    "university"
]

found_education = []

for keyword in education_keywords:
    if keyword in text:
        found_education.append(keyword)

# -----------------------------
# Certification Detection
# -----------------------------

certification_keywords = [
    "certificate",
    "certification",
    "nptel",
    "course",
    "training"
]

found_certifications = []

for keyword in certification_keywords:
    if keyword in text:
        found_certifications.append(keyword)

# -----------------------------
# Experience Detection
# -----------------------------

experience_keywords = [
    "internship",
    "intern",
    "experience",
    "developer",
    "training"
]

found_experience = []

for keyword in experience_keywords:
    if keyword in text:
        found_experience.append(keyword)

# -----------------------------
# Project Detection
# -----------------------------

project_keywords = [
    "project",
    "system",
    "application",
    "platform"
]

found_projects = []

for keyword in project_keywords:
    if keyword in text:
        found_projects.append(keyword)

# -----------------------------
# Word Count
# -----------------------------

words = re.findall(r'\b\w+\b', resume)
word_count = len(words)

# -----------------------------
# Resume Score
# -----------------------------

score = 0

if len(found_skills) >= 5:
    score += 30
elif len(found_skills) >= 3:
    score += 20
elif len(found_skills) >= 1:
    score += 10

if len(found_education) >= 2:
    score += 20
elif len(found_education) >= 1:
    score += 10

if len(found_certifications) >= 2:
    score += 20
elif len(found_certifications) >= 1:
    score += 10

if len(found_experience) >= 2:
    score += 15
elif len(found_experience) >= 1:
    score += 10

if len(found_projects) >= 2:
    score += 15
elif len(found_projects) >= 1:
    score += 10

# -----------------------------
# Suggestions
# -----------------------------

suggestions = []

if len(found_skills) < 5:
    suggestions.append("Add more relevant technical skills.")

if len(found_certifications) == 0:
    suggestions.append("Add relevant certifications.")

if len(found_experience) == 0:
    suggestions.append("Add internship or project experience.")

if len(found_projects) == 0:
    suggestions.append("Add academic or personal projects.")

if word_count < 100:
    suggestions.append("Add more details about your skills and achievements.")

if not suggestions:
    suggestions.append("Resume contains good basic information.")

# -----------------------------
# Display Result
# -----------------------------

print("=" * 60)
print("             AI-BASED RESUME TEXT ANALYZER")
print("=" * 60)

print("\n--- RESUME STATISTICS ---")
print("Word Count:", word_count)

print("\n--- TECHNICAL SKILLS ---")
print(found_skills if found_skills else "No skills detected")

print("\n--- EDUCATION ---")
print(found_education if found_education else "No education details detected")

print("\n--- CERTIFICATIONS ---")
print(found_certifications if found_certifications else "No certifications detected")

print("\n--- EXPERIENCE ---")
print(found_experience if found_experience else "No experience detected")

print("\n--- PROJECT INFORMATION ---")
print(found_projects if found_projects else "No projects detected")

print("\n--- RESUME SCORE ---")
print("Score:", score, "/ 100")

if score >= 80:
    print("Rating: Excellent ⭐")
elif score >= 60:
    print("Rating: Good 👍")
elif score >= 40:
    print("Rating: Average 🙂")
else:
    print("Rating: Needs Improvement ⚠️")

print("\n--- SUGGESTIONS ---")

for suggestion in suggestions:
    print("•", suggestion)

print("\n" + "=" * 60)
```
